### 5. Hiển thị Kết Quả & Dashboard Báo Cáo Trực Tiếp Trên Colab

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

# 1. Hiển thị Physical Dashboard
if os.path.exists('reports/physical_dashboard.png'):
    img = Image.open('reports/physical_dashboard.png')
    plt.figure(figsize=(15, 9))
    plt.imshow(img)
    plt.axis('off')
    plt.title('Physical & Speed Performance Dashboard', fontsize=16)
    plt.show()

# 2. Hiển thị 2D Positional Heatmaps
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
if os.path.exists('reports/team_1_heatmap.png'):
    axes[0].imshow(Image.open('reports/team_1_heatmap.png'))
    axes[0].axis('off')
    axes[0].set_title('Team 1 Heatmap', fontsize=14)
if os.path.exists('reports/team_2_heatmap.png'):
    axes[1].imshow(Image.open('reports/team_2_heatmap.png'))
    axes[1].axis('off')
    axes[1].set_title('Team 2 Heatmap', fontsize=14)
plt.show()

# 3. Hiển thị Passing Networks
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
if os.path.exists('reports/team_1_passing_network.png'):
    axes[0].imshow(Image.open('reports/team_1_passing_network.png'))
    axes[0].axis('off')
    axes[0].set_title('Team 1 Passing Network', fontsize=14)
if os.path.exists('reports/team_2_passing_network.png'):
    axes[1].imshow(Image.open('reports/team_2_passing_network.png'))
    axes[1].axis('off')
    axes[1].set_title('Team 2 Passing Network', fontsize=14)
plt.show()
```
